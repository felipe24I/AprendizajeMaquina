#**Crear archivo dashboard_regresion.py para visualización con Streamlit**

Crea un dashboard en Streamlit para comparar tres modelos de regresión (RandomForest, LinearRegression y ElasticNet) que cargue las predicciones desde archivos .npy (ytest.npy, y_pred_RF.npy, y_pred_LR.npy, y_pred_Elastic.npy). El dashboard debe incluir: (1) una tabla comparativa de métricas (MAE, MSE, R², MAPE) con highlight de los mejores valores, (2) gráficos interactivos de barras para comparar métricas seleccionables, (3) scatter plots de predicciones vs valores reales con línea de referencia, (4) visualización de hiperparámetros óptimos por modelo, y (5) un sidebar con opciones. Usa Streamlit, Plotly y Matplotlib, con diseño responsive y estilo profesional. Guarda el código en dashboard_regresion.py con %%writefile.

```python
%%writefile dashboard_regresion.py
import streamlit as st
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px
import numpy as np

# Cargar predicciones y ytest desde archivos .npy
ytest = np.load('/content/drive/MyDrive/ytest.npy')
y_pred_RF = np.load('/content/drive/MyDrive/y_pred_RF.npy')
y_pred_BR = np.load('/content/drive/MyDrive/y_pred_BR.npy')
y_pred_EN = np.load('/content/drive/MyDrive/y_pred_EN.npy')
# Configuración de la página
st.set_page_config(page_title="Análisis de Modelos de Regresión", layout="wide")
st.title("Dashboard Comparativo: Modelos de Regresión")

# Datos de métricas
metrics = {
    "RandomForest": {"MAE": 15979.313093, "MSE": 699819188.706366, "R2": 0.901626, "MAPE": 0.099262},
    "BayesianRidge": {"MAE": 20231.059614, "MSE": 989816624.360191, "R2": 0.860860, "MAPE": 0.121041},
    "ElasticNet": {"MAE": 20253.525920, "MSE": 990640719.396923, "R2": 0.860744, "MAPE": 0.121382}
}

# --- Página Principal ---
st.header("Resumen Comparativo")
df_metrics = pd.DataFrame(metrics).T
st.dataframe(df_metrics.style.highlight_min(axis=0), height=200)

# Gráfico de métricas comparativas
metric_option = st.selectbox("Selecciona una métrica para comparar:", ["MAE", "MSE", "R2", "MAPE"])
fig = px.bar(df_metrics, y=metric_option, title=f"Comparación de {metric_option} entre Modelos")
st.plotly_chart(fig, use_container_width=True)

# --- Pestañas por Modelo ---
tab1, tab2, tab3 = st.tabs(["RandomForest", "BayesianRidge", "ElasticNet"])

with tab1:
    st.header("RandomForest")
    col1, col2 = st.columns(2)
    with col1:
        st.subheader("Mejores Hiperparámetros")
        st.json({
            "n_estimators": 96,
            "max_depth": 35,
            "min_samples_split": 2,
            "min_samples_leaf": 1,
            "max_features": "log2"
        })
    with col2:
        st.subheader("Métricas")
        st.write(df_metrics.loc["RandomForest"].to_frame().T)

    fig, ax = plt.subplots()
    ax.scatter(ytest, y_pred_RF, alpha=0.5)
    ax.plot([ytest.min(), ytest.max()], [ytest.min(), ytest.max()], 'r--')
    ax.set_title("Predicciones vs Valores Reales (RandomForest)")
    st.pyplot(fig)

with tab2:
    st.header("BayesianRidge")
    col1, col2 = st.columns(2)
    with col1:
        st.subheader("Mejores Hiperparámetros")
        st.json({
            "alpha_1": 0.1,
            "alpha_2": 0.000261,
            "lambda_1": 0.000194,
            "lambda_2": 0.020204
        })
    with col2:
        st.subheader("Métricas")
        st.write(df_metrics.loc["BayesianRidge"].to_frame().T)

    fig, ax = plt.subplots()
    ax.scatter(ytest, y_pred_BR, alpha=0.5)
    ax.plot([ytest.min(), ytest.max()], [ytest.min(), ytest.max()], 'r--')
    ax.set_title("Predicciones vs Valores Reales (BayesianRidge)")
    st.pyplot(fig)

with tab3:
    st.header("ElasticNet")
    col1, col2 = st.columns(2)
    with col1:
        st.subheader("Mejores Parámetros")
        st.json({
            "alpha": 0.0001,
            "l1_ratio": 0.9
        })
    with col2:
        st.subheader("Métricas")
        st.write(df_metrics.loc["ElasticNet"].to_frame().T)

    fig, ax = plt.subplots()
    ax.scatter(ytest, y_pred_EN, alpha=0.5)
    ax.plot([ytest.min(), ytest.max()], [ytest.min(), ytest.max()], 'r--')
    ax.set_title("Predicciones vs Valores Reales (ElasticNet)")
    st.pyplot(fig)

# --- Sidebar ---
with st.sidebar:
    st.header("Opciones")
    show_code = st.checkbox("Mostrar código del dashboard")
    if show_code:
        st.code("""
        # Código disponible en GitHub:
        # https://github.com/tu_usuario/tu_repo
        """)

st.markdown("---")
st.caption("Dashboard creado por Felipe Idárraga | Datos actualizados: 2025")
```
